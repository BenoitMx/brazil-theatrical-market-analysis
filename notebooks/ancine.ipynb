```python
import pandas as pd

query = """
SELECT *
FROM `Ancine.films_w1_w4`
"""

df_ancine = pd.read_gbq(
    query,
    project_id="benoit-project-488815"
)

df_ancine.head()
```

```python
df_ancine.isnull().mean().sort_values(ascending=False).head(20)
```

```python
df_ancine = df_ancine[
    (df_ancine["week_retention_w2_pct"].between(-100, 150)) &
    (df_ancine["revenue_per_screen_w1"].between(8000, 60000))
]
```

```python
ancine_numeric = df_ancine.drop(["release_date_clean", "title", "genre", 'production_countries', "nationality", "distributor", "distributor_origin"], axis=1)
ancine_numeric.head()
```

```python
print(ancine_numeric.shape)
ancine_numeric = ancine_numeric.dropna()
print(ancine_numeric.shape)
```

```python
ancine_numeric = ancine_numeric.dropna()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
ancine_scaled = scaler.fit_transform(ancine_numeric)

ancine_scaled = pd.DataFrame(
    ancine_scaled,
    columns=ancine_numeric.columns,
    index=ancine_numeric.index
)
ancine_scaled.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

kmeans_10_labels = KMeans(n_clusters=10, n_init=10)


kmeans_10_labels.fit(ancine_scaled)

# Add predictions to the original dataframe
ancine_scaled["kmeans_10_labels_pred"] = kmeans_10_labels.labels_

ancine_scaled.head()
```

```python
ancine_scaled.groupby("kmeans_10_labels_pred").count()
```

```python
inertias = []
n_k = range(1, 11)

# for each number of centroids
for k in n_k:  
    model = KMeans(n_clusters=k, random_state=0, n_init=10)
    model.fit(ancine_scaled)
    inertias.append(model.inertia_)

#Graph
import plotly.express as px
fig = px.scatter(inertias)
fig.show()
```

```python
ancine_scaled = ancine_scaled.drop(["kmeans_10_labels_pred"], axis=1)
```

```python
kmeans_4_labels = KMeans(n_clusters=4, n_init=10)

kmeans_4_labels.fit(ancine_scaled)

# Add predictions to the original dataframe
df_ancine["label"] = kmeans_4_labels.labels_

df_ancine.head()
```

```python
fig_scaled = px.scatter_3d(df_ancine,
                           x="screens_w1",
                           y='revenue_per_screen_w1',
                           z='week_retention_w2_pct',
                           color="label",
                           width=500,
                           height=500)
fig_scaled.show()
```

```python
releases_clusters = {}

for num_cluster in df_ancine["label"].unique():
    releases_clusters[num_cluster] = df_ancine[
        df_ancine["label"] == num_cluster
    ].sample(15)
    
releases_clusters
```

```python
df_ancine.to_csv(
    "ancine_clusters.csv",
    index=False
)
```
